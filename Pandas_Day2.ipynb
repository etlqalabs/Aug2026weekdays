```python
# Filtering ( where clause in SQL)
# Get me all the employees who nbelongs to deptno = 30 and salary > 3000
df = pd.read_csv("emp.csv")
filter1  = df['deptno'] == 30
filter2  = df['salary'] > 3000
df[filter1&filter2]
```

```python
# Get me all the employees who nbelongs to deptno = 30 or salary > 3000
df = pd.read_csv("emp.csv")
filter1  = df['deptno'] == 30
filter2  = df['salary'] > 3000
df[filter1|filter2]
```
