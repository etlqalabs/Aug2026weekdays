# Group by

```python
# Aggregate functions (sum,min,max,count,avg
```

```python
#Q1 . Count the employees in each department
# expected o/p : deptno,count_employees
# SQL : select deptno,count(deptno) from emp(df) group by deptno

aggr_data = df.groupby('deptno')['eno'].count()
aggr_data
```

```python
type(aggr_data)
```

```python
#convert the series in to dataframe
aggr_data = aggr_data.reset_index()
aggr_data
```

```python
type(aggr_data)
```

```python
#rename eno column in to count_employees
aggr_data.rename(columns={"eno":"count_employees"},inplace=True)
```

```python
aggr_data
```

```python
#Q2. get me total count of employees, max salary, min salary,total salary in each department
# SQl: select count(eno) as employees_count,max(salary) as max_salary,min(salary) as min_salary,sum(salary) as total_salary 
# group by deptno

agg_data = df.groupby('deptno')['salary'].agg(['count','max','min','sum'])
```

```python
type(agg_data)
```

```python
agg_data
```

```python
#Flatten the data to add indexes
agg_data = agg_data.reset_index()
```

```python
agg_data.rename(columns={'count':"employee_count",'max':"max_salary","min":"min_salary","sum":"total_salary"},inplace=True)
```

```python
agg_data
```

```python
# Write this summary data in to emp_summary.csv
agg_data.to_csv("emp_summary.csv",index=False)
```

```python
#Q3. Get me employees count in each location and within each deptno

# expected o/p : 
#location deptno emp_count
#Bangalore 10     2
#Bangalore 20     5
#Chennai   20     5
#Chennai   30     8
# SQL equivalent  : select location,deptno ,count(eno) as emp_count from emp group by location,deptno

data = df.groupby(['location','deptno'])['eno'].count().reset_index()
data.rename(columns = {"eno":"emp_count"},inplace=True)
data
```

```python
#Q4. Get me total salary in each department and expect for location ="mumbai" and total salary must be 

#SQL : select deptno,sum(salary) as total_sal from emp where location !="mumbai" group by deptno having sum(salary)>50K

filter_location = df['location'] != "Mumbai"
df_filtered = df[filter_location]
df_agg = df_filtered.groupby('deptno')['salary'].sum()
df_agg1 = df_agg.reset_index()
```

```python
# Way 1
having_fileter = df_agg1['salary'] >50000
df_agg1[having_fileter]
```

```python
# way2 (Lambda function)
lamda_fil = df_agg.apply(lambda x:x>50000)
df_agg[lamda_fil].reset_index()
```
