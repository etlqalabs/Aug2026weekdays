```python
import pandas as pd
```

# merge ( sql joins)

```python
df_emp = pd.read_csv("data/employees.csv")
```

```python
df_emp.head()
```

```python
df_dept = pd.read_csv("data/department.csv")
```

```python
df_dept.head()
```

# inner join

```python
# SQL : select * from empartment as emp inner join department as dept on emp.deptno = dept.deptid

ans_df = df_emp.merge(df_dept,left_on='deptno',right_on='deptid',how='inner')
```

```python
ans_df
```

# left join

```python
# SQL : select * from employees as emp left join department as dept on emp.deptno = dept.deptid

ans_df = df_emp.merge(df_dept,left_on='deptno',right_on='deptid',how='left')
```

```python
ans_df
```

# right join

```python
# SQL : select * from employees as emp right join department as dept on emp.deptno = dept.deptid

ans_df = df_emp.merge(df_dept,left_on='deptno',right_on='deptid',how='right')
```

```python
ans_df
```

```python
# same can be acheved using left join by changing the dataframes sequence
# SQL : select * from department as dept left join employees as emp on emp.deptno = dept.deptid

ans_df = df_dept.merge(df_emp,left_on='deptid',right_on='deptno',how='left')
```

```python
ans_df
```

# full join

```python
# SQL : select * from employees as emp full join department as dept on emp.deptno = dept.deptid

ans_df = df_emp.merge(df_dept,left_on='deptno',right_on='deptid',how='outer')
```

```python
ans_df
```

# Concat()

```python
# concat() used to concatenate either alongs rows(axis = 0 ) or along columns (axis =1)
# along axis = 0 ( along rows)
#df1(5 rows)
#df2(4 rows)

# along axis = 1 ( along columns)
#df1(5 rows) df2(4 rows)

```

```python
pd.concat([df_emp,df_dept],axis=0)
```

```python
pd.concat([df_emp,df_dept],axis=1)
```

```python
pd.concat([df_dept,df_emp],axis=1)
```
